```python
class Chocolate:
    def __init__(self, price, additive):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        self.price = float(price)
        self.additive = str(additive)

    def show_my_chocolate(self):
        if self.additive:
            print (f'Шоколадный батончик, добавка: {self.additive}')
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return (f'Шоколадный батончик, {self.price} рублей')


chocolate1 = Chocolate(4, "Orex")

chocolate2=Chocolate(15,"Cocos")
chocolate3=Chocolate(10,"Nuga")
print(chocolate1)
print(chocolate2)
print(chocolate3)
chocolate1.show_my_chocolate()
chocolate2.show_my_chocolate()
chocolate3.show_my_chocolate()
```
